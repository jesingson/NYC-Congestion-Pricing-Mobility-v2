```
# ---------------------------------------------------------------------
# Validate Bus calendar-expanded temporal fields
# ---------------------------------------------------------------------
# Lightweight contract check. The Bus spatiotemporal input is large, so we
# validate the required fields and inspect a small bounded sample rather than
# scanning the full partitioned dataset.

BUS_TEMPORAL_DIAGNOSTIC_SAMPLE_ROWS = 1_000
BUS_REQUIRED_TEMPORAL_COLUMNS = [
    "date",
    "day_of_week",
    "hour",
    "daypart",
    "daypart_order",
    "temporal_bucket",
    "temporal_bucket_order",
    "observation_sequence_id",
    "pre_post_cp",
    "temporal_grain_type",
    "date_assignment_method",
]

expected_temporal_bucket_values = [
    "weekday_overnight",
    "weekday_am_peak",
    "weekday_midday",
    "weekday_pm_peak",
    "weekday_evening",
    "weekend_overnight",
    "weekend_am_peak",
    "weekend_midday",
    "weekend_pm_peak",
    "weekend_evening",
]

bus_columns_df = duckdb.sql(
    f"DESCRIBE SELECT * FROM read_parquet('{bus_spatiotemporal_glob}') LIMIT 0"
).df()

bus_columns = set(bus_columns_df["column_name"].tolist())
missing_bus_temporal_columns = [
    column_name for column_name in BUS_REQUIRED_TEMPORAL_COLUMNS
    if column_name not in bus_columns
]

if missing_bus_temporal_columns:
    display(bus_columns_df)
    raise ValueError(
        "Bus spatiotemporal input is missing required temporal contract columns: "
        f"{missing_bus_temporal_columns}"
    )

valid_buckets_sql = ", ".join([f"'{bucket}'" for bucket in expected_temporal_bucket_values])

bus_temporal_validation_df = duckdb.sql(f"""
WITH bus_sample AS (
    SELECT
        date,
        day_of_week,
        hour,
        daypart,
        daypart_order,
        temporal_bucket,
        temporal_bucket_order,
        observation_sequence_id,
        pre_post_cp,
        temporal_grain_type,
        date_assignment_method
    FROM read_parquet('{bus_spatiotemporal_glob}')
    LIMIT {BUS_TEMPORAL_DIAGNOSTIC_SAMPLE_ROWS}
)
SELECT
    COUNT(*) AS sampled_row_count,
    MIN(CAST(date AS DATE)) AS min_date,
    MAX(CAST(date AS DATE)) AS max_date,
    COUNT(DISTINCT CAST(date AS DATE)) AS distinct_dates,
    COUNT(DISTINCT day_of_week) AS distinct_day_of_week_values,
    COUNT(DISTINCT hour) AS distinct_hour_values,
    COUNT(DISTINCT temporal_bucket) AS distinct_temporal_buckets,
    SUM(CASE WHEN date IS NULL THEN 1 ELSE 0 END) AS missing_date_rows,
    SUM(CASE WHEN day_of_week IS NULL THEN 1 ELSE 0 END) AS missing_day_of_week_rows,
    SUM(CASE WHEN hour IS NULL THEN 1 ELSE 0 END) AS missing_hour_rows,
    SUM(CASE WHEN daypart IS NULL THEN 1 ELSE 0 END) AS missing_daypart_rows,
    SUM(CASE WHEN daypart_order IS NULL THEN 1 ELSE 0 END) AS missing_daypart_order_rows,
    SUM(CASE WHEN temporal_bucket IS NULL OR temporal_bucket NOT IN ({valid_buckets_sql}) THEN 1 ELSE 0 END) AS invalid_temporal_bucket_rows,
    SUM(CASE WHEN temporal_bucket_order IS NULL THEN 1 ELSE 0 END) AS missing_temporal_bucket_order_rows,
    SUM(CASE WHEN observation_sequence_id IS NULL THEN 1 ELSE 0 END) AS missing_observation_sequence_rows,
    SUM(CASE WHEN pre_post_cp IS NULL THEN 1 ELSE 0 END) AS missing_pre_post_cp_rows,
    SUM(CASE WHEN temporal_grain_type IS NULL THEN 1 ELSE 0 END) AS missing_temporal_grain_type_rows,
    SUM(CASE WHEN date_assignment_method IS NULL THEN 1 ELSE 0 END) AS missing_date_assignment_method_rows
FROM bus_sample
""").df()

bus_temporal_validation_df["sample_limit"] = BUS_TEMPORAL_DIAGNOSTIC_SAMPLE_ROWS
bus_temporal_validation_df["available_column_count"] = len(bus_columns)

print("Bus temporal field check completed against the current 1.2.2 Bus contract.")
display(bus_temporal_validation_df)
```

Findings\. The Traffic dataset contains observations spanning January 2023 through February 2026 but only includes 714 distinct dates across the study period\. Given that the calendar window contains more than 1,100 days, this indicates that Traffic does not provide continuous daily observations\. Instead, the dataset appears to consist of periodic traffic\-count collections conducted on selected dates throughout the study period\.

```
traffic_gap_df = duckdb.sql(f"""
WITH dates AS (
    SELECT DISTINCT date
    FROM read_parquet('{traffic_spatiotemporal_path}')
),
date_gaps AS (
    SELECT
        date,
        LAG(date) OVER (ORDER BY date) AS previous_date
    FROM dates
)
SELECT
    date,
    previous_date,
    date - previous_date AS gap_days
FROM date_gaps
WHERE previous_date IS NOT NULL
ORDER BY gap_days DESC
LIMIT 20
""").df()

display(traffic_gap_df)
```

Findings\. The date\-gap analysis confirmed the presence of substantial gaps between traffic collection periods, including intervals of 75, 49, 27, and 25 days between observed dates\. These gaps are too large to be explained by occasional missing observations and instead suggest that the underlying traffic program consists of periodic roadway count studies rather than continuously monitored traffic sensors\. Because these gaps reflect the structure of the source data rather than missing records, no temporal interpolation will be performed\. Traffic observations will be retained only for dates that were actually collected by the source program\.

### Build calendar expansion

First, create a compact calendar table for the study period\. This table gives us every valid calendar date, along with its year, month, and day of week, so we can map Bus monthly weekday profiles onto the matching dates\.

```
# ---------------------------------------------------------------------
# 1.2.4.2 Harmonize Bus Temporal Representation
# ---------------------------------------------------------------------

bus_expansion_calendar_df = pd.DataFrame({
    "date": pd.date_range(
        start=STUDY_START_DATE,
        end=STUDY_END_DATE,
        freq="D"
    )
})

bus_expansion_calendar_df["year"] = (
    bus_expansion_calendar_df["date"].dt.year
)

bus_expansion_calendar_df["month"] = (
    bus_expansion_calendar_df["date"].dt.month
)

bus_expansion_calendar_df["day_of_week"] = (
    bus_expansion_calendar_df["date"].dt.day_name()
)

bus_expansion_calendar_df.head()
```

Next, register the calendar as a DuckDB view\. This allows us to join the Bus monthly weekday/hour profiles to matching calendar dates without loading the full Bus dataset into Pandas\.

```
duckdb.register(
    "bus_expansion_calendar",
    bus_expansion_calendar_df
)
```

### Build Harmonized Bus Dataset

Next, create the final harmonized Bus mobility table\. Rather than writing expanded route\-segment\-level records, we aggregate directly to Taxi Zone × Date × Temporal Bucket, which is the grain needed for downstream integration\.

```
if REBUILD_HARMONIZED_BUS or not harmonized_bus_path.exists():

    print("Building harmonized Bus mobility table...")

    duckdb.sql(f"""
        COPY (
            SELECT
                c.date::DATE AS date,
                EXTRACT(YEAR FROM c.date)::INTEGER AS year,
                EXTRACT(MONTH FROM c.date)::INTEGER AS month,
                STRFTIME(c.date, '%A') AS day_of_week,
                b.temporal_bucket,
                CASE
                    WHEN c.date >= DATE '2025-01-05'
                    THEN 'post_cp'
                    ELSE 'pre_cp'
                END AS pre_post_cp,
                b.bus_taxi_zone_id::INTEGER AS taxi_zone_id,

                AVG(b."Average Road Speed") AS avg_bus_speed,
                AVG(b."Average Travel Time") AS avg_bus_travel_time,

                COALESCE(
                    STDDEV_SAMP(b."Average Road Speed"),
                    0
                ) AS bus_speed_stddev,

                COALESCE(
                    STDDEV_SAMP(b."Average Travel Time"),
                    0
                ) AS bus_travel_time_stddev,
                SUM(b."Bus Trip Count") AS bus_trip_count,
                COUNT(*) AS bus_segment_observation_count,
                COUNT(DISTINCT b.bus_taxi_zone_assignment_method) AS bus_assignment_method_count

            FROM read_parquet('{bus_spatiotemporal_glob}') AS b
            INNER JOIN bus_expansion_calendar AS c
                ON b.year_1 = c.year
                AND b.month_1 = c.month
                AND b.day_of_week = c.day_of_week

            WHERE
                b.bus_taxi_zone_id IS NOT NULL

            GROUP BY
                c.date,
                b.temporal_bucket,
                b.bus_taxi_zone_id
        )
        TO '{harmonized_bus_path}'
        (FORMAT PARQUET, COMPRESSION ZSTD)
    """)

    print(f"Saved harmonized Bus table to: {harmonized_bus_path}")

else:
    print(f"Harmonized Bus table already exists: {harmonized_bus_path}")
```

### Unrelated Newark Sanity Check

```
# Quick sanity check: LocationID 1 / Newark Airport across modalities
# ---------------------------------------------------------------------
# Newark Airport is a valid TLC Taxi Zone, but should generally not appear
# in NYC roadway, Bus, or Subway spatial assignments. The TLC contract now
# uses the canonical pickup-zone field `taxi_zone_id` rather than the old
# intermediate `pickup_zone_id` name.

NEWARK_AIRPORT_LOCATION_ID = 1
NEWARK_SANITY_SAMPLE_ROWS = 100_000

newark_modality_specs = [
    {
        "dataset": "traffic",
        "source": str(traffic_spatiotemporal_path),
        "candidate_zone_fields": ["taxi_zone_id", "location_id"],
    },
    {
        "dataset": "bus",
        "source": str(bus_spatiotemporal_glob),
        "candidate_zone_fields": ["taxi_zone_id", "bus_taxi_zone_id"],
    },
    {
        "dataset": "subway",
        "source": str(subway_spatiotemporal_glob),
        "candidate_zone_fields": ["taxi_zone_id"],
    },
    {
        "dataset": "tlc_final_contract",
        "source": str(tlc_final_contract_glob),
        "candidate_zone_fields": ["taxi_zone_id"],
    },
]

newark_sanity_rows = []

for spec in newark_modality_specs:
    dataset = spec["dataset"]
    source = spec["source"]

    schema_df = duckdb.sql(f"""
        DESCRIBE SELECT *
        FROM read_parquet('{source}')
        LIMIT 0
    """).df()

    available_columns = set(schema_df["column_name"].tolist())
    zone_field = next(
        (
            candidate_field
            for candidate_field in spec["candidate_zone_fields"]
            if candidate_field in available_columns
        ),
        None,
    )

    if zone_field is None:
        newark_sanity_rows.append({
            "dataset": dataset,
            "source": source,
            "location_field": None,
            "sampled_row_count": None,
            "newark_sample_row_count": None,
            "min_sample_date": None,
            "max_sample_date": None,
            "sample_limit": NEWARK_SANITY_SAMPLE_ROWS,
            "status": "Review: no expected zone field found",
        })
        continue

    zone_field_sql = f'"{zone_field}"' if not zone_field.replace("_", "").isalnum() else zone_field
    date_expr = "CAST(date AS DATE)" if "date" in available_columns else "NULL"

    newark_counts_df = duckdb.sql(f"""
        WITH modality_sample AS (
            SELECT *
            FROM read_parquet('{source}')
            LIMIT {NEWARK_SANITY_SAMPLE_ROWS}
        )
        SELECT
            COUNT(*) AS sampled_row_count,
            SUM(CASE WHEN CAST({zone_field_sql} AS INTEGER) = {NEWARK_AIRPORT_LOCATION_ID} THEN 1 ELSE 0 END)
                AS newark_sample_row_count,
            MIN(CASE WHEN CAST({zone_field_sql} AS INTEGER) = {NEWARK_AIRPORT_LOCATION_ID} THEN {date_expr} ELSE NULL END)
                AS min_sample_date,
            MAX(CASE WHEN CAST({zone_field_sql} AS INTEGER) = {NEWARK_AIRPORT_LOCATION_ID} THEN {date_expr} ELSE NULL END)
                AS max_sample_date
        FROM modality_sample
    """).df()

    result = newark_counts_df.iloc[0].to_dict()
    result.update({
        "dataset": dataset,
        "source": source,
        "location_field": zone_field,
        "sample_limit": NEWARK_SANITY_SAMPLE_ROWS,
        "status": "Expected TLC-only presence" if dataset == "tlc_final_contract" else "Expected no / minimal presence",
    })
    newark_sanity_rows.append(result)

newark_airport_sanity_df = pd.DataFrame(newark_sanity_rows)[
    [
        "dataset",
        "location_field",
        "sampled_row_count",
        "newark_sample_row_count",
        "min_sample_date",
        "max_sample_date",
        "sample_limit",
        "status",
        "source",
    ]
]

display(newark_airport_sanity_df)
```

### Aggregated Traffic Coverage Inspection

Before moving on, let's quantify the coverage of the harmonized Traffic table\. Because Traffic observations are collected on a sampled schedule rather than a continuous daily schedule, we expect some gaps in the final Taxi Zone × Date × Temporal Bucket panel\. This QA helps us understand how much of the potential Traffic observation space is actually represented in the harmonized dataset\.

```
traffic_coverage_df = duckdb.sql(f"""
WITH traffic_summary AS (
    SELECT
        COUNT(DISTINCT date) AS observed_dates,
        COUNT(DISTINCT taxi_zone_id) AS observed_zones,
        COUNT(DISTINCT temporal_bucket) AS observed_buckets,
        COUNT(*) AS observed_records
    FROM read_parquet('{harmonized_traffic_path}')
)

SELECT
    observed_dates,
    observed_zones,
    observed_buckets,
    observed_records,

    observed_dates
        * observed_zones
        * observed_buckets
        AS theoretical_records,

    ROUND(
        100.0 * observed_records
        / (
            observed_dates
            * observed_zones
            * observed_buckets
        ),
        2
    ) AS percent_complete
FROM traffic_summary
""").df()

display(traffic_coverage_df)
```

Overall completeness can sometimes hide localized coverage gaps\. Next, let's identify Taxi Zones that have limited Traffic coverage and determine whether certain areas of the city are systematically underrepresented in the Traffic dataset\.

### Create Harmonized Subway Mobility Table

Now that we have confirmed the Subway dataset is already temporally and spatially standardized, we can create the final Subway mobility table\. This step aggregates the granular station\-level observations to the project's canonical Taxi Zone × Date × Temporal Bucket grain, sums the core Subway mobility measures, and retains lightweight diagnostic counts for downstream QA and troubleshooting\.

The final Subway mobility table is created by grouping observations by Taxi Zone, date, and temporal bucket\. We derive the standard temporal fields from date, recalculate the congestion\-pricing period indicator for consistency with the other harmonized mobility tables, and aggregate ridership, transfers, observation counts, and distinct station\-complex counts\.

```
# Create harmonized Subway mobility table

duckdb.sql(f"""
    COPY (
        SELECT
            date::DATE AS date,
            EXTRACT(YEAR FROM date)::INTEGER AS year,
            EXTRACT(MONTH FROM date)::INTEGER AS month,
            STRFTIME(date, '%A') AS day_of_week,
            temporal_bucket,
            CASE
                WHEN date >= DATE '2025-01-05'
                THEN 'post_cp'
                ELSE 'pre_cp'
            END AS pre_post_cp,
            taxi_zone_id::INTEGER AS taxi_zone_id,

            SUM(ridership) AS subway_ridership,
            SUM(transfers) AS subway_transfers,
            COUNT(*) AS subway_observation_count,
            COUNT(DISTINCT station_complex_id) AS subway_station_complex_count

        FROM read_parquet('{subway_spatiotemporal_glob}')
        WHERE
            taxi_zone_id IS NOT NULL
            AND date >= DATE '{STUDY_START_DATE}'
            AND date <= DATE '{STUDY_END_DATE}'

        GROUP BY
            date,
            temporal_bucket,
            taxi_zone_id
    )
    TO '{harmonized_subway_path}'
    (FORMAT PARQUET);
""")

print(f"Harmonized Subway mobility table written to: {harmonized_subway_path}")
```

```
# ---------------------------------------------------------------------
# Build Harmonized TLC Dataset from the final 1.2.3 contract
# ---------------------------------------------------------------------

print("Final TLC contract input:", tlc_final_contract_glob)
print("Harmonized TLC output path:", harmonized_tlc_path)

if REBUILD_HARMONIZED_TLC or not harmonized_tlc_path.exists():

    tlc_contract_files = sorted(tlc_final_contract_glob.parent.glob("*.parquet"))

    assert tlc_contract_files, (
        "No final TLC contract files found. "
        "Run 1.2.3 final TLC contract build before 1.2.4 harmonization."
    )

    print(
        "Building harmonized TLC table directly from final Taxi/FHVHV contract "
        f"({len(tlc_contract_files)} monthly chunks)...",
        flush=True
    )

    con = duckdb.connect()
    con.execute("PRAGMA threads=1")
    con.execute("PRAGMA memory_limit='8GB'")

    con.execute(f"""
        COPY (
            WITH contract_rows AS (
                SELECT
                    modality,
                    taxi_zone_id,
                    CAST(date AS DATE) AS date,
                    year,
                    month,
                    day_of_week,
                    daypart,
                    daypart_order,
                    temporal_bucket,
                    temporal_bucket_order,
                    observation_sequence_id,
                    pre_post_cp,
                    trip_count,
                    avg_trip_distance,
                    avg_trip_duration_seconds,
                    avg_trip_speed_mph,
                    distinct_dropoff_zone_count,
                    source_od_row_count,
                    spatial_assignment_method,
                    spatial_assignment_source,
                    spatial_assignment_confidence,
                    canonical_geography
                FROM read_parquet('{tlc_final_contract_glob.as_posix()}')
                WHERE CAST(date AS DATE) BETWEEN DATE '{STUDY_START_DATE}'
                    AND DATE '{STUDY_END_DATE}'
            )
            SELECT
                taxi_zone_id,
                date,
                year,
                month,
                day_of_week,
                daypart,
                daypart_order,
                temporal_bucket,
                temporal_bucket_order,
                observation_sequence_id,
                pre_post_cp,

                SUM(CASE WHEN modality = 'taxi' THEN trip_count ELSE 0 END)
                    AS taxi_trip_count,
                MAX(CASE WHEN modality = 'taxi' THEN avg_trip_distance END)
                    AS taxi_avg_trip_distance,
                MAX(CASE WHEN modality = 'taxi' THEN avg_trip_duration_seconds END)
                    AS taxi_avg_trip_duration,
                MAX(CASE WHEN modality = 'taxi' THEN avg_trip_speed_mph END)
                    AS taxi_avg_trip_speed,
                MAX(CASE WHEN modality = 'taxi' THEN distinct_dropoff_zone_count END)
                    AS taxi_distinct_dropoff_zone_count,
                SUM(CASE WHEN modality = 'taxi' THEN source_od_row_count ELSE 0 END)
                    AS taxi_source_od_row_count,

                SUM(CASE WHEN modality = 'fhvhv' THEN trip_count ELSE 0 END)
                    AS fhvhv_trip_count,
                MAX(CASE WHEN modality = 'fhvhv' THEN avg_trip_distance END)
                    AS fhvhv_avg_trip_distance,
                MAX(CASE WHEN modality = 'fhvhv' THEN avg_trip_duration_seconds END)
                    AS fhvhv_avg_trip_duration,
                MAX(CASE WHEN modality = 'fhvhv' THEN avg_trip_speed_mph END)
                    AS fhvhv_avg_trip_speed,
                MAX(CASE WHEN modality = 'fhvhv' THEN distinct_dropoff_zone_count END)
                    AS fhvhv_distinct_dropoff_zone_count,
                SUM(CASE WHEN modality = 'fhvhv' THEN source_od_row_count ELSE 0 END)
                    AS fhvhv_source_od_row_count,

                COUNT(DISTINCT modality) AS tlc_modality_count,
                COUNT(*) AS tlc_contract_row_count,
                STRING_AGG(DISTINCT spatial_assignment_method, ', ' ORDER BY spatial_assignment_method)
                    AS spatial_assignment_method,
                STRING_AGG(DISTINCT spatial_assignment_source, ', ' ORDER BY spatial_assignment_source)
                    AS spatial_assignment_source,
                STRING_AGG(DISTINCT spatial_assignment_confidence, ', ' ORDER BY spatial_assignment_confidence)
                    AS spatial_assignment_confidence,
                STRING_AGG(DISTINCT canonical_geography, ', ' ORDER BY canonical_geography)
                    AS canonical_geography
            FROM contract_rows
            GROUP BY
                taxi_zone_id,
                date,
                year,
                month,
                day_of_week,
                daypart,
                daypart_order,
                temporal_bucket,
                temporal_bucket_order,
                observation_sequence_id,
                pre_post_cp
            ORDER BY
                taxi_zone_id,
                date,
                temporal_bucket_order
        )
        TO '{harmonized_tlc_path.as_posix()}'
        (FORMAT PARQUET)
    """)

    con.close()

    print(f"Harmonized TLC mobility table written to: {harmonized_tlc_path}")

else:

    print(f"Using existing harmonized TLC mobility table: {harmonized_tlc_path}")
```

### Create Harmonized Bridge & Tunnel Mobility Table

Now we’ll create the final Bridge & Tunnel mobility table by aggregating hourly crossing observations into Facility × Direction × Date × Temporal Bucket records\. Payment method and vehicle class are consolidated, while directional crossing volume is preserved as the core mobility signal\.

```
# Create harmonized Bridge & Tunnel mobility table

if REBUILD_HARMONIZED_BRIDGE_TUNNEL or not harmonized_bridge_tunnel_path.exists():

    print("Building harmonized Bridge & Tunnel mobility table...")

    duckdb.sql(f"""
        COPY (
            SELECT
                date::DATE AS date,
                EXTRACT(YEAR FROM date)::INTEGER AS year,
                EXTRACT(MONTH FROM date)::INTEGER AS month,
                STRFTIME(date, '%A') AS day_of_week,
                temporal_bucket,
                CASE
                    WHEN date >= DATE '2025-01-05'
                    THEN 'post_cp'
                    ELSE 'pre_cp'
                END AS pre_post_cp,

                facility_id::INTEGER AS facility_id,
                facility,
                direction,

                SUM(traffic_count) AS bridge_tunnel_volume,
                COUNT(*) AS bridge_tunnel_observation_count

            FROM read_parquet('{bridge_tunnel_temporal_path}')
            WHERE
                date >= DATE '{STUDY_START_DATE}'
                AND date <= DATE '{STUDY_END_DATE}'
                AND facility_id IS NOT NULL
                AND facility IS NOT NULL
                AND direction IS NOT NULL

            GROUP BY
                date,
                temporal_bucket,
                facility_id,
                facility,
                direction
        )
        TO '{harmonized_bridge_tunnel_path}'
        (FORMAT PARQUET, COMPRESSION ZSTD)
    """)

    print(f"Saved harmonized Bridge & Tunnel table to: {harmonized_bridge_tunnel_path}")

else:
    print(f"Harmonized Bridge & Tunnel table already exists: {harmonized_bridge_tunnel_path}")
```
